 ## 📋 2표본 비율 검정 핵심 정리



 ### 1. 2표본 비율 검정 개념과 사용 상황



 | 구분 | 내용 | 예시 |
 |------|------|------|
 | **목적** | 두 집단의 성공 비율 비교 | A안 vs B안 전환율 비교 |
 | **독립변수** | 2개 집단 (범주형) | A그룹, B그룹 |
 | **종속변수** | 이진 결과 (성공/실패) | 클릭/미클릭, 구매/미구매 |
 | **핵심 질문** | "두 집단의 비율이 같은가?" | "A안과 B안의 클릭률이 다른가?" |
 | **Python 함수** | `proportions_ztest()` | `proportions_ztest(counts, nobs)` |
 > * 2표본 비율: **두 독립 집단**의 비율(%, 전환율) 비교


 ### 3. 검정 방향 선택
| 검정 유형 | 가설 | 사용 시기 | 실무 적용 |
|----------|------|----------|-----------|
| **양측 검정** | H₀: p₁ = p₂ vs H₁: p₁ ≠ p₂ | 방향 예측 없음 | **A/B 테스트** (기본) |
| **단측 검정** | H₀: p₁ ≤ p₂ vs H₁: p₁ > p₂ | 사전에 방향 명확 | **임상시험** (일반적) |

**Python 함수 사용법:**
- 양측 검정: `proportions_ztest(counts, nobs, alternative='two-sided')` (기본값)
- 단측 검정: `proportions_ztest(counts, nobs, alternative='larger')` 또는 `fisher_exact(table, alternative='greater')`


 ### 4. 검정 가정사항

 | 가정사항 | 기준 | 확인 방법 | 위반시 대안 |
 |---------|------|----------|-------------|
 | **표본 크기** | 성공/실패 각각 ≥ 5 | 빈도 확인 | Fisher's exact test |
 | **독립성** | 두 집단이 서로 독립 | 연구 설계 확인 | McNemar 검정 (대응표본) |
 | **무작위성** | 무작위 추출 | 표본 추출 방법 확인 | - |



 ### 5. 효과 크기와 신뢰구간
 통계적 유의성(p-value)은 차이의 존재 여부만 알려주지만, 효과 크기와 신뢰구간은 **차이의 실질적 중요성과 확실성**을 나타냅니다.

 | 측정치 | 의미 | 실무 활용 | 예시 해석 |
 |--------|------|----------|----------|
 | **비율 차이** | 절대적 차이 (점추정) | 비용/수익 계산 | 5%p → "10,000명당 500건 증가" |
 | **비율비 (RR)** | 상대적 개선율 | 경영진 보고 | 1.33 → "B가 A의 1.33배" 또는 "33% 증가" |
 | **95% 신뢰구간** | 불확실성 범위 (구간추정) | 리스크 평가 | [2%p~8%p] → "최악 2%p, 최선 8%p" |

**비율비(Rate Ratio) 해석:**
- RR < 1.0: B가 A보다 낮음
- RR = 1.0: 차이 없음
- RR > 1.0: B가 A보다 높음
- 예: RR = 1.33 → B가 A의 1.33배 (33% 증가)
 > 💡 **실무 팁**: 효과 크기(얼마나 큰가?) + 신뢰구간(얼마나 확실한가?) 모두 확인

 **효과 크기 vs 신뢰구간:**

 ```plain
 효과 크기:   차이가 5%p (단일 값)
 신뢰구간:   95% 신뢰수준에서 실제 차이는 2%p~8%p 사이로 추정

 → 같은 5%p 개선이어도
    CI [4%p~6%p]: 확실함 (표본 충분)
    CI [-2%p~12%p]: 불확실함 (표본 부족, 실제로는 마이너스일 수도)
 ```

 ### 6. 전체 분석 프로세스

 1. **데이터 탐색**
    - 각 그룹의 성공/실패 횟수 확인
    - 비율 계산 및 시각화

 2. **가정 검정**
    - 표본 크기 확인 (성공/실패 각 ≥ 5)
    - Fisher's exact test 필요성 판단

 3. **검정 수행**
    - z-검정 또는 Fisher's exact test
    - p-value 확인

 4. **효과 크기 및 신뢰구간 계산**
    - 비율 차이 (절대적 크기)
    - 비율비 (상대적 개선율)
    - 95% 신뢰구간 (불확실성 범위)

 5. **결과 보고**
    - "B안이 A안보다 전환율 5%p 높음 (33% 개선)"
    - "95% CI: [2%p~8%p], 최악의 경우에도 2%p 개선 추정"

 ### 💡 핵심 요약
 - **언제**: 두 집단의 비율(전환율, 클릭률) 비교 (A/B 테스트)
 - **가정**: 독립성, 충분한 표본 크기 (성공/실패 각 ≥ 5)
 - **검정**: 양측 검정 기본, 일측은 사전 방향 예측시만
 - **해석**: p-value(유의성) + 비율 차이(절대 크기) + 비율비(상대 개선율) + 신뢰구간(확실성)
 - **주의**: 통계적 유의성 ≠ 실질적 중요성
 - **실무**: 신뢰구간으로 리스크 평가 필수

 ## 도우미 함수

````python
def check_sample_size_assumption(count1, n1, count2, n2):
    """
    2표본 비율 검정의 표본 크기 가정 확인
    
    z-검정(정규근사)을 사용하기 위한 최소 표본 크기 조건을 확인합니다.
    각 그룹에서 성공 횟수와 실패 횟수가 모두 5 이상이어야 합니다.
    
    Parameters
    ----------
    count1 : int
        그룹 1의 성공 횟수
    n1 : int
        그룹 1의 총 시도 횟수
    count2 : int
        그룹 2의 성공 횟수
    n2 : int
        그룹 2의 총 시도 횟수
    
    Returns
    -------
    bool
        z-검정 사용 가능 여부
    """
    print("\n[표본 크기 가정 확인]")
    print("-"*60)
    
    # 각 그룹의 성공/실패 횟수
    success1 = count1
    failure1 = n1 - count1
    success2 = count2
    failure2 = n2 - count2
    
    print(f"그룹 1: 성공 {success1}회, 실패 {failure1}회")
    print(f"그룹 2: 성공 {success2}회, 실패 {failure2}회")
    
    # 모든 값이 5 이상인지 확인
    all_counts = [success1, failure1, success2, failure2]
    min_count = min(all_counts)
    
    print(f"최소 빈도: {min_count}")
    
    if min_count >= 5:
        print("✅ 모든 빈도 ≥ 5")
        print("   → z-검정(정규근사) 사용 적합")
        return True
    else:
        print(f"⚠️ 일부 빈도 < 5 (최소값: {min_count})")
        print("   → Fisher's exact test 사용 권장")
        return False


def calculate_effect_sizes(count1, n1, count2, n2):
    """
    효과 크기 계산
    
    비율 차이, 비율비(Rate Ratio), 신뢰구간을 계산합니다.
    
    Parameters
    ----------
    count1 : int
        그룹 1의 성공 횟수 (일반적으로 대조군 또는 기준 그룹)
    n1 : int
        그룹 1의 총 시도 횟수
    count2 : int
        그룹 2의 성공 횟수 (일반적으로 실험군 또는 비교 그룹)
    n2 : int
        그룹 2의 총 시도 횟수
    
    Returns
    -------
    dict
        효과 크기 정보를 담은 딕셔너리
        - p1: 그룹 1의 비율
        - p2: 그룹 2의 비율
        - diff: 비율 차이 (p2 - p1)
        - rr: 비율비 (p2 / p1)
        - ci1, ci2: 각 그룹의 95% 신뢰구간
        - ci_diff: 비율 차이의 95% 신뢰구간
    
    Notes
    -----
    비율 차이와 비율비는 항상 (그룹2 - 그룹1) 또는 (그룹2 / 그룹1)로 계산됩니다.
    함수 호출 시 인자 순서에 주의하세요.
    """
    
    # 비율 계산
    p1 = count1 / n1
    p2 = count2 / n2
    
    # 비율 차이
    diff = p2 - p1
    
    # 비율비 (Rate Ratio)
    rr = p2 / p1 if p1 > 0 else None
    
    # 각 그룹의 95% 신뢰구간
    ci1 = proportion_confint(count1, n1, method='wilson')
    ci2 = proportion_confint(count2, n2, method='wilson')
    
    # 비율 차이의 95% 신뢰구간 (근사)
    se_diff = np.sqrt(p1*(1-p1)/n1 + p2*(1-p2)/n2)
    ci_diff = (diff - 1.96*se_diff, diff + 1.96*se_diff)
    
    return {
        'p1': p1,
        'p2': p2,
        'diff': diff,
        'rr': rr,
        'ci1': ci1,
        'ci2': ci2,
        'ci_diff': ci_diff
    }
````

 ## 실습 1: 이메일 마케팅 A/B 테스트 (양측검정 및 가정 만족)



 ### 문제: 두 가지 이메일 디자인의 클릭률 비교

 이커머스 회사에서 두 가지 이메일 디자인을 테스트했습니다.
- 디자인 A: 기존 디자인 (단순한 텍스트 중심)
- 디자인 B: 새로운 디자인 (이미지와 CTA 버튼 강조)

각 디자인을 무작위로 고객에게 발송하고 클릭 여부를 측정했습니다.

[연구 질문]
"두 디자인의 클릭률에 차이가 있는가?"

````python
print("\n" + "="*60)
print("실습 1: 이메일 마케팅 A/B 테스트 - 2표본 비율 검정")
print("="*60)

# 데이터 설정
design_A_clicks = 142      # 디자인 A 클릭 수
design_A_total = 1000      # 디자인 A 발송 수

design_B_clicks = 178      # 디자인 B 클릭 수
design_B_total = 1000      # 디자인 B 발송 수

# 데이터프레임으로 정리
data_summary = pd.DataFrame({
    '디자인': ['A (기존)', 'B (신규)'],
    '발송수': [design_A_total, design_B_total],
    '클릭수': [design_A_clicks, design_B_clicks],
    '클릭률(%)': [design_A_clicks/design_A_total*100, design_B_clicks/design_B_total*100]
})

print(f"\n[데이터 요약]")
display(data_summary)
````

````python
# 시각화
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# 비율 비교 막대 그래프
categories = ['디자인 A', '디자인 B']
click_rates = [design_A_clicks/design_A_total*100, design_B_clicks/design_B_total*100]
axes[0].bar(categories, click_rates, color=['#ff9999', '#66b3ff'], alpha=0.7)
axes[0].set_title('디자인별 클릭률 비교')
axes[0].set_ylabel('클릭률 (%)')
axes[0].grid(True, alpha=0.3, axis='y')
for i, v in enumerate(click_rates):
    axes[0].text(i, v + 0.5, f'{v:.1f}%', ha='center')

# 성공/실패 스택 막대 그래프
success = [design_A_clicks, design_B_clicks]
failure = [design_A_total - design_A_clicks, design_B_total - design_B_clicks]
x = np.arange(len(categories))
axes[1].bar(x, success, label='클릭', color='#66b3ff', alpha=0.8)
axes[1].bar(x, failure, bottom=success, label='미클릭', color='lightgray', alpha=0.8)
axes[1].set_title('디자인별 클릭/미클릭 분포')
axes[1].set_xticks(x)
axes[1].set_xticklabels(categories)
axes[1].set_ylabel('고객 수')
axes[1].legend()
axes[1].grid(True, alpha=0.3, axis='y')

# 신뢰구간 포함 비교
effects = calculate_effect_sizes(design_A_clicks, design_A_total, 
                                 design_B_clicks, design_B_total)
ci_lower = [effects['ci1'][0]*100, effects['ci2'][0]*100]
ci_upper = [effects['ci1'][1]*100, effects['ci2'][1]*100]
yerr = [[click_rates[i] - ci_lower[i] for i in range(2)],
        [ci_upper[i] - click_rates[i] for i in range(2)]]
axes[2].bar(categories, click_rates, yerr=yerr, capsize=10,
           color=['#ff9999', '#66b3ff'], alpha=0.7)
axes[2].set_title('디자인별 클릭률 (95% CI)')
axes[2].set_ylabel('클릭률 (%)')
axes[2].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
````

````python
# =============================================================================
# 2표본 비율 검정 수행
# =============================================================================

print("\n[2표본 비율 검정]")
print("-"*60)

# 가설 설정
print("H₀: 두 디자인의 클릭률이 같다 (p₁ = p₂)")
print("H₁: 두 디자인의 클릭률이 다르다 (p₁ ≠ p₂)")
print("유의수준: α = 0.05")

# -----------------------------------------------------------------------------
# 1. 가정 검정
# -----------------------------------------------------------------------------
is_valid = check_sample_size_assumption(design_A_clicks, design_A_total,
                                       design_B_clicks, design_B_total)

# -----------------------------------------------------------------------------
# 2. 검정 수행
# -----------------------------------------------------------------------------
print("\n[검정 결과]")
print("-"*60)

counts = np.array([design_A_clicks, design_B_clicks])
nobs = np.array([design_A_total, design_B_total])

# z-검정 수행
z_stat, p_value = proportions_ztest(counts, nobs, alternative='two-sided')

print(f"z-통계량: {z_stat:.4f}")
print(f"p-value: {p_value:.4f}")

# -----------------------------------------------------------------------------
# 3. 효과 크기 계산
# -----------------------------------------------------------------------------
print("\n[효과 크기]")
print("-"*60)

effects = calculate_effect_sizes(design_A_clicks, design_A_total,
                                 design_B_clicks, design_B_total)

print(f"디자인 A 클릭률: {effects['p1']*100:.2f}%")
print(f"디자인 B 클릭률: {effects['p2']*100:.2f}%")
print(f"비율 차이: {effects['diff']*100:.2f}%p")
print(f"비율비 (RR): {effects['rr']:.3f}")
if effects['rr'] > 1:
    print(f"  → 디자인 B가 A의 {effects['rr']:.2f}배 ({(effects['rr']-1)*100:.1f}% 증가)")
elif effects['rr'] < 1:
    print(f"  → 디자인 B가 A보다 낮음 ({(1-effects['rr'])*100:.1f}% 감소)")
else:
    print(f"  → 두 디자인 동일")
print(f"비율 차이 95% CI: [{effects['ci_diff'][0]*100:.2f}%p, {effects['ci_diff'][1]*100:.2f}%p]")

# -----------------------------------------------------------------------------
# 4. 통계적 결론
# -----------------------------------------------------------------------------
print("\n[결론]")
if p_value < 0.05:
    print(f"✅ p-value({p_value:.4f}) < 0.05 → 귀무가설 기각")
    print("   두 디자인의 클릭률에 유의한 차이가 있음")
else:
    print(f"❌ p-value({p_value:.4f}) ≥ 0.05 → 귀무가설 채택")
    print("   두 디자인의 클릭률에 유의한 차이가 없음")
````

````python
# =============================================================================
# 비즈니스 의사결정
# =============================================================================

if p_value < 0.05:
    print("\n[비즈니스 인사이트]")
    print("-"*50)
    print(f"• 디자인 B의 클릭률이 디자인 A보다 {effects['diff']*100:.1f}%p 높음")
    print(f"• 디자인 B가 A 대비 {(effects['rr']-1)*100:.1f}% 증가 ({effects['rr']:.2f}배)")
    print(f"• 95% 신뢰구간: [{effects['ci_diff'][0]*100:.1f}%p ~ {effects['ci_diff'][1]*100:.1f}%p]")
    print(f"• 월 10만명 발송 시 약 {abs(effects['diff'] * 100000):.0f}건 추가 클릭 예상")
    print("• 디자인 B를 전체 적용 권장")
else:
    print("\n[비즈니스 인사이트]")
    print("-"*50)
    print(f"• 두 디자인 간 클릭률 차이 없음 (차이: {effects['diff']*100:.1f}%p)")
    print("• 기존 디자인 A 유지 권장")
    print("• 제목이나 발송 시간대 등 다른 요소 테스트 고려")
````

 ### 문제: 신약과 위약의 치료 성공률 비교 (작은 표본)

희귀질환 치료를 위한 신약의 소규모 임상시험을 진행했습니다.
환자 모집의 어려움으로 인해 표본 크기가 작습니다.

- 신약 그룹: 신규 개발 약물 투여
- 위약 그룹: 플라시보(가짜약) 투여

[연구 질문]
"신약이 위약보다 치료 성공률이 높은가?"

````python
print("\n" + "="*60)
print("실습 2: 신약 임상시험 - Fisher's Exact Test")
print("="*60)


# 데이터 설정 (작은 표본)
drug_success = 9       # 신약 그룹 치료 성공
drug_total = 12        # 신약 그룹 총 환자

placebo_success = 3    # 위약 그룹 치료 성공
placebo_total = 10     # 위약 그룹 총 환자

# 데이터프레임으로 정리
trial_summary = pd.DataFrame({
    '그룹': ['신약', '위약'],
    '총환자수': [drug_total, placebo_total],
    '치료성공': [drug_success, placebo_success],
    '치료실패': [drug_total - drug_success, placebo_total - placebo_success],
    '성공률(%)': [drug_success/drug_total*100, placebo_success/placebo_total*100]
})

print(f"\n[데이터 요약]")
display(trial_summary)
````

````python
# 시각화
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# 성공률 비교
categories = ['신약', '위약']
success_rates = [drug_success/drug_total*100, placebo_success/placebo_total*100]
axes[0].bar(categories, success_rates, color=['#4ECDC4', '#ff9999'], alpha=0.7)
axes[0].set_title('치료 성공률 비교')
axes[0].set_ylabel('성공률 (%)')
axes[0].set_ylim(0, 100)
axes[0].grid(True, alpha=0.3, axis='y')
for i, v in enumerate(success_rates):
    axes[0].text(i, v + 2, f'{v:.1f}%', ha='center')

# 성공/실패 스택 그래프
success = [drug_success, placebo_success]
failure = [drug_total - drug_success, placebo_total - placebo_success]
x = np.arange(len(categories))
axes[1].bar(x, success, label='성공', color='#4ECDC4', alpha=0.8)
axes[1].bar(x, failure, bottom=success, label='실패', color='lightcoral', alpha=0.8)
axes[1].set_title('치료 결과 분포')
axes[1].set_xticks(x)
axes[1].set_xticklabels(categories)
axes[1].set_ylabel('환자 수')
axes[1].legend()
axes[1].grid(True, alpha=0.3, axis='y')

# 분할표 히트맵
contingency = [[drug_success, drug_total - drug_success],
               [placebo_success, placebo_total - placebo_success]]
im = axes[2].imshow(contingency, cmap='Blues', aspect='auto')
axes[2].set_xticks([0, 1])
axes[2].set_yticks([0, 1])
axes[2].set_xticklabels(['성공', '실패'])
axes[2].set_yticklabels(['신약', '위약'])
axes[2].set_title('2x2 분할표')
for i in range(2):
    for j in range(2):
        text = axes[2].text(j, i, contingency[i][j],
                          ha="center", va="center", color="black", fontsize=14)
plt.colorbar(im, ax=axes[2])

plt.tight_layout()
plt.show()
````

````python
# =============================================================================
# Fisher's Exact Test 수행
# =============================================================================

print("\n[Fisher's Exact Test]")
print("-"*60)

# 가설 설정 (단측)
print("H₀: 신약의 성공률 ≤ 위약")
print("H₁: 신약의 성공률 > 위약 (단측 검정)")
print("유의수준: α = 0.05")

# -----------------------------------------------------------------------------
# 1. 가정 검정
# -----------------------------------------------------------------------------
# 데이터프레임 순서와 일치시킴: 신약 먼저, 위약 나중
is_valid = check_sample_size_assumption(drug_success, drug_total,
                                       placebo_success, placebo_total)

# -----------------------------------------------------------------------------
# 2. Fisher's Exact Test 수행
# -----------------------------------------------------------------------------
print("\n⚠️ 표본 크기 가정을 만족하지 않으므로 Fisher's exact test를 수행합니다.")
print("\n[Fisher's Exact Test 결과]")
print("-"*60)

# 분할표 생성 (행: 그룹, 열: 성공/실패)
contingency_table = [[drug_success, drug_total - drug_success],      # 신약
                     [placebo_success, placebo_total - placebo_success]]  # 위약

print("\n[2x2 분할표 구조]")
print("           성공    실패")
print(f"신약        {drug_success}      {drug_total - drug_success}")
print(f"위약        {placebo_success}       {placebo_total - placebo_success}")

# Fisher's exact test (단측)
# alternative='greater'는 첫 번째 행(신약)의 성공 오즈가 두 번째 행(위약)보다 크다는 대립가설을 의미
odds_ratio, p_value = fisher_exact(contingency_table, alternative='greater') #단측검정

print(f"오즈비 (Odds Ratio): {odds_ratio:.4f}")
# 오즈비 해석 
if odds_ratio > 1:
    print(f"→ 신약의 성공 오즈가 위약의 {odds_ratio:.2f}배")
elif odds_ratio < 1:
    print(f"→ 신약의 성공 오즈가 위약의 {odds_ratio:.2f}배 (위약이 더 좋음)")
else:
    print(f"→ 두 그룹의 성공 오즈가 동일")
    
print(f"p-value (단측): {p_value:.4f}")

# -----------------------------------------------------------------------------
# 3. 효과 크기 계산
# -----------------------------------------------------------------------------
print("\n[효과 크기]")
print("-"*60)

# 인자 순서: (대조군, 대조군 합계, 실험군, 실험군 합계)
# 위약(대조군) → 신약(실험군)
effects = calculate_effect_sizes(placebo_success, placebo_total,
                                 drug_success, drug_total)

print(f"위약 성공률: {effects['p1']*100:.2f}% ({placebo_success}/{placebo_total})")
print(f"신약 성공률: {effects['p2']*100:.2f}% ({drug_success}/{drug_total})")
print(f"비율 차이: {effects['diff']*100:.2f}%p")
print(f"비율비 (RR): {effects['rr']:.3f}")
if effects['rr'] > 1:
    print(f"  → 신약이 위약의 {effects['rr']:.2f}배 ({(effects['rr']-1)*100:.1f}% 증가)")
elif effects['rr'] < 1:
    print(f"  → 신약이 위약보다 낮음 (위약의 {effects['rr']*100:.0f}%)")
else:
    print(f"  → 두 그룹 동일")
print(f"오즈비 (OR): {odds_ratio:.3f}")

# -----------------------------------------------------------------------------
# 4. 통계적 결론
# -----------------------------------------------------------------------------
print("\n[결론]")
if p_value < 0.05:
    print(f"✅ p-value({p_value:.4f}) < 0.05 → 귀무가설 기각")
    print("   신약이 위약보다 치료 성공률이 유의하게 높음")
else:
    print(f"❌ p-value({p_value:.4f}) ≥ 0.05 → 귀무가설 채택")
    print("   신약이 위약보다 우수하다는 증거가 불충분함")
````

````python
# %%
# =============================================================================
# 임상적 해석
# =============================================================================

if p_value < 0.05:
    print("\n[임상적 해석]")
    print("-"*50)
    print(f"• 신약의 성공률이 위약보다 {effects['diff']*100:.1f}%p 높음")
    print(f"• 신약이 위약의 {effects['rr']:.1f}배 효과")
    print(f"• 오즈비 {odds_ratio:.1f} (신약의 성공 오즈가 {odds_ratio:.0f}배)")
    print("• 신약의 우월성이 통계적으로 입증됨 (단측 검정)")
else:
    print("\n[임상적 해석]")
    print("-"*50)
    print(f"• 신약의 우월성 입증 실패 (단측 p={p_value:.3f})")
    print(f"• 신약 성공률 {effects['p2']*100:.0f}% vs 위약 {effects['p1']*100:.0f}%")
    print(f"• 표본 크기 부족 가능성 (n={drug_total + placebo_total})")
    print("• 더 큰 규모의 연구 또는 용량 조정 고려")
````

## 실습 3: 모바일 앱 푸시알림 전략 (대응표본 - McNemar)

### 문제: 푸시알림 개선이 앱 재방문율을 향상시켰는가?

[배경]
모바일 쇼핑앱에서 푸시알림 전략을 개선하고, **동일한 사용자 150명**을 대상으로
개선 전후 1주일간의 앱 재방문 여부를 추적했습니다.

- 개선 전: 일반적인 상품 알림
- 개선 후: 개인화된 추천 알림

[연구 질문]
"푸시알림 개선이 앱 재방문율을 향상시켰는가?"

**중요**: 동일한 사용자의 전후 비교이므로 **대응표본**입니다!

````python
print("\n" + "="*60)
print("실습 3: 푸시알림 전략 개선 - McNemar 검정")
print("="*60)

# 데이터 설정 (McNemar 검정용 2x2 분할표)
# 행: 개선 전 상태, 열: 개선 후 상태
both_visit = 45          # 전후 모두 재방문
before_only = 12         # 전만 재방문 (개선 후 이탈)
after_only = 28          # 후만 재방문 (개선 후 유입)
neither = 65             # 전후 모두 미방문

mcnemar_table = np.array([[both_visit, before_only],
                          [after_only, neither]])

# 데이터프레임으로 정리
print(f"\n[2x2 분할표]")
contingency_df = pd.DataFrame(mcnemar_table, 
                              columns=['개선 후: 재방문', '개선 후: 미방문'],
                              index=['개선 전: 재방문', '개선 전: 미방문'])
display(contingency_df)

# 요약 통계
total_users = both_visit + before_only + after_only + neither
before_visit = both_visit + before_only
after_visit = both_visit + after_only

summary_df = pd.DataFrame({
    '구분': ['개선 전 재방문율', '개선 후 재방문율', '순증가'],
    '값': [f'{before_visit}/{total_users} ({before_visit/total_users*100:.1f}%)',
           f'{after_visit}/{total_users} ({after_visit/total_users*100:.1f}%)',
           f'+{after_only - before_only}명 (+{(after_only - before_only)/total_users*100:.1f}%p)']
})

print(f"\n[데이터 요약]")
display(summary_df)
````

````python
# 시각화
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# 전후 비교 막대그래프
categories = ['개선 전', '개선 후']
visit_rates = [before_visit/total_users*100, after_visit/total_users*100]
axes[0].bar(categories, visit_rates, color=['#ff9999', '#4ECDC4'], alpha=0.7)
axes[0].set_title('푸시알림 개선 전후 재방문율')
axes[0].set_ylabel('재방문율 (%)')
axes[0].set_ylim(0, 60)
axes[0].grid(True, alpha=0.3, axis='y')
for i, v in enumerate(visit_rates):
    axes[0].text(i, v + 1, f'{v:.1f}%', ha='center')

# 변화 유형별 사용자 분포
change_types = ['전후\n모두\n재방문', '전만\n재방문\n(이탈)', '후만\n재방문\n(유입)', '전후\n모두\n미방문']
counts = [both_visit, before_only, after_only, neither]
colors_change = ['#4ECDC4', '#FFB6B6', '#90EE90', '#D3D3D3']
axes[1].bar(change_types, counts, color=colors_change, alpha=0.7)
axes[1].set_title('사용자 변화 유형별 분포')
axes[1].set_ylabel('사용자 수 (명)')
axes[1].grid(True, alpha=0.3, axis='y')
for i, v in enumerate(counts):
    axes[1].text(i, v + 2, f'{v}명', ha='center')

# 2x2 분할표 히트맵
im = axes[2].imshow(mcnemar_table, cmap='Blues', aspect='auto', vmin=0, vmax=70)
axes[2].set_xticks([0, 1])
axes[2].set_yticks([0, 1])
axes[2].set_xticklabels(['개선 후\n재방문', '개선 후\n미방문'])
axes[2].set_yticklabels(['개선 전\n재방문', '개선 전\n미방문'])
axes[2].set_title('2x2 분할표 히트맵')

# 히트맵에 숫자 표시
for i in range(2):
    for j in range(2):
        text = axes[2].text(j, i, mcnemar_table[i, j],
                          ha="center", va="center", 
                          color="white" if mcnemar_table[i, j] > 35 else "black",
                          fontsize=16, fontweight='bold')
plt.colorbar(im, ax=axes[2])

plt.tight_layout()
plt.show()
````

````python
# =============================================================================
# McNemar 검정 수행 (개선된 버전)
# =============================================================================

print("\n[McNemar 검정]")
print("-"*60)

# 가설 설정
print("H₀: 푸시알림 개선 전후 재방문율이 같다")
print("H₁: 푸시알림 개선 전후 재방문율이 다르다")
print("유의수준: α = 0.05")
print("\n※ 동일한 사용자의 전후 비교이므로 McNemar 검정 사용")

# McNemar 검정 수행
from statsmodels.stats.contingency_tables import mcnemar
from statsmodels.stats.proportion import proportion_confint
import numpy as np

result = mcnemar(mcnemar_table, exact=False, correction=True)

print("\n[검정 결과]")
print("-"*60)
print(f"카이제곱 통계량: {result.statistic:.4f}")
print(f"p-value: {result.pvalue:.4f}")

# =============================================================================
# 효과 크기 분석 (신뢰구간 포함)
# =============================================================================

print("\n[효과 크기 및 신뢰구간]")
print("-"*60)

# 1. 불일치 쌍 분석 (McNemar의 핵심)
discordant_pairs = before_only + after_only
print(f"불일치 쌍 (b+c): {discordant_pairs}명")
print(f"  • 개선 후 이탈 (b): {before_only}명")
print(f"  • 개선 후 유입 (c): {after_only}명")
print(f"  • 순증가: {after_only - before_only}명")

# 2. 재방문율 변화
before_visit_rate = before_visit / total_users
after_visit_rate = after_visit / total_users

print(f"\n재방문율:")
print(f"  • 개선 전: {before_visit_rate*100:.1f}% ({before_visit}/{total_users})")
print(f"  • 개선 후: {after_visit_rate*100:.1f}% ({after_visit}/{total_users})")
print(f"  • 증가: {(after_visit_rate - before_visit_rate)*100:.1f}%p")

# 3. 각 시점의 재방문율 신뢰구간
ci_before = proportion_confint(before_visit, total_users, method='wilson')
ci_after = proportion_confint(after_visit, total_users, method='wilson')

print(f"\n[재방문율 95% 신뢰구간]")
print(f"  • 개선 전 재방문율: [{ci_before[0]*100:.1f}%, {ci_before[1]*100:.1f}%]")
print(f"  • 개선 후 재방문율: [{ci_after[0]*100:.1f}%, {ci_after[1]*100:.1f}%]")

# 4. 비율 차이의 신뢰구간 (McNemar 전용)
diff = (after_only - before_only) / total_users

# McNemar 검정에서 비율 차이의 표준오차
se_diff = np.sqrt((before_only + after_only)) / total_users

# 95% 신뢰구간
ci_diff_lower = diff - 1.96 * se_diff
ci_diff_upper = diff + 1.96 * se_diff

print(f"\n[재방문율 차이 95% 신뢰구간]")
print(f"  • 비율 차이: {diff*100:.1f}%p")
print(f"  • 95% CI: [{ci_diff_lower*100:.1f}%p, {ci_diff_upper*100:.1f}%p]")

# 신뢰구간 해석
if ci_diff_lower > 0:
    print(f"    → 95% 신뢰수준에서 최소 {ci_diff_lower*100:.1f}%p 개선 확신")
elif ci_diff_upper < 0:
    print(f"    → 95% 신뢰수준에서 오히려 감소한 것으로 추정")
else:
    print(f"    → 95% 신뢰구간이 0을 포함 (효과 불확실)")

# 6. 통계적 결론
print("\n[결론]")
print("-"*60)
if result.pvalue < 0.05:
    print(f"✅ p-value({result.pvalue:.4f}) < 0.05 → 귀무가설 기각")
    print("   푸시알림 개선 전후 재방문율에 유의한 차이가 있음")
else:
    print(f"❌ p-value({result.pvalue:.4f}) ≥ 0.05 → 귀무가설 채택")
    print("   푸시알림 개선 전후 재방문율에 유의한 차이가 없음")
````

````python
# =============================================================================
# 비즈니스 인사이트
# =============================================================================

if result.pvalue < 0.05:
    print("\n[비즈니스 인사이트]")
    print("-"*50)
    print(f"• 개인화 알림으로 재방문율 {(after_visit - before_visit)/total_users*100:.1f}%p 증가")
    print(f"  - 95% CI: [{ci_diff_lower*100:.1f}%p, {ci_diff_upper*100:.1f}%p]")
    print(f"• 순증가 {after_only - before_only}명 (유입 {after_only}명 - 이탈 {before_only}명)")
    print(f"• 개선 후가 개선 전의 {after_visit/before_visit:.2f}배 효과")
    print(f"• 최소 개선 효과: {ci_diff_lower*100:.1f}%p (95% 신뢰수준)")
    print(f"• 전체 사용자 대상 개인화 알림 전면 적용 권장")
    print(f"• 예상 효과: 10만명 기준 약 {(after_visit - before_visit)/total_users * 100000:.0f}명 재방문 증가")
    print(f"  - 최소: {ci_diff_lower * 100000:.0f}명, 최대: {ci_diff_upper * 100000:.0f}명")
else:
    print("\n[비즈니스 인사이트]")
    print("-"*50)
    print(f"• 푸시알림 개선 효과가 통계적으로 유의하지 않음")
    print(f"• 순증가 {after_only - before_only}명 (유입 {after_only}명 - 이탈 {before_only}명)")
    print(f"• 95% CI: [{ci_diff_lower*100:.1f}%p, {ci_diff_upper*100:.1f}%p] (0 포함)")
    print(f"• 개인화 알고리즘 재검토 필요")
    print(f"• 알림 타이밍, 빈도, 메시지 내용 등 다른 요소 테스트 권장")
````

## 📝 실습 문제 3: 웹사이트 디자인 개편 효과


### 배경
전자상거래 웹사이트를 개편하고, **동일한 사용자 120명**을 대상으로
개편 전후의 구매 여부를 추적했습니다.

### 연구 질문
**"웹사이트 개편이 구매율을 향상시켰는가?"**

[배경]
전자상거래 웹사이트 디자인을 전면 개편했습니다.
- 개편 전: 기존 UI/UX
- 개편 후: 모바일 최적화 + 간소화된 결제 프로세스

[데이터 - 2x2 분할표]
```plain
                개편 후 구매    개편 후 미구매
개편 전 구매         30              8         (38명)
개편 전 미구매       25              57        (82명)
                   (55명)          (65명)
```

````python
# 데이터
both_purchase = 30       # 전후 모두 구매
before_only = 8          # 전만 구매
after_only = 25          # 후만 구매
neither = 57             # 전후 모두 미구매

mcnemar_table = np.array([[both_purchase, before_only],
                          [after_only, neither]])

print(f"\n[2x2 분할표]")
contingency_df = pd.DataFrame(mcnemar_table, 
                              columns=['개편 후: 구매', '개편 후: 미구매'],
                              index=['개편 전: 구매', '개편 전: 미구매'])
display(contingency_df)

# 요약 통계 계산
total_users = both_purchase + before_only + after_only + neither
before_purchase = both_purchase + before_only
after_purchase = both_purchase + after_only

summary_df = pd.DataFrame({
    '구분': ['개편 전 구매율', '개편 후 구매율', '순증가'],
    '값': [f'{before_purchase}/{total_users} ({before_purchase/total_users*100:.1f}%)',
           f'{after_purchase}/{total_users} ({after_purchase/total_users*100:.1f}%)',
           f'+{after_only - before_only}명 (+{(after_only - before_only)/total_users*100:.1f}%p)']
})

print(f"\n[데이터 요약]")
display(summary_df)
````
