 ## 📋 카이제곱검정 핵심 정리



 ### 1. 카이제곱검정 종류와 사용 상황

 | 종류 | 사용 상황 | 예시 | Python 함수 |
 |------|-----------|------|------------|
 | **적합도 검정** | 관측빈도와 기대빈도 비교 | 주사위 공정성 검정 | `chisquare()` |
 | **독립성 검정** | 두 범주형 변수간 관계성 | 성별과 제품선호도 관계 | `chi2_contingency()` |
 | **동질성 검정** | 여러 집단의 분포 비교 | 지역별 투표 성향 비교 | `chi2_contingency()` |
 > - 카이제곱: 범주형 변수의 **관계성/분포** 검정

 ### 2. 카이제곱검정 가정사항
 | 가정사항 | 기준 | 위반시 대안 |
 |---------|------|------------|
 | **최소 기대빈도** | 모든 셀 ≥ 5 (5미만 셀이 20% 이하면 허용) | Fisher's exact test |
 | **범주형 변수** | 명목/서열 척도 | 연속형은 구간화 필요 |
 | **독립 관측** | 각 관측치는 한 셀에만 속함 | 다른 통계방법 고려 |

 ### 3. 효과 크기 측정 - 상황별 선택
 | 측정치 | 언제 사용? | 해석 기준 | 계산식 |
 |--------|-----------|----------|--------|
 | **Cramér's V** | **모든 크기의 표**<br>(가장 범용적) | df=1: 0.1(작음), 0.3(중간), 0.5(큼)<br>df=2: 0.07(작음), 0.21(중간), 0.35(큼)<br>df=3: 0.06(작음), 0.17(중간), 0.29(큼) | $V = \sqrt{\frac{\chi^2}{n \times df}}$ |
 > 자유도(df) = min(행-1, 열-1)



 ### 4. 조정 잔차 분석 - 구체적 패턴 파악
 | 구분 | 조정 잔차 (Adjusted Residuals) |
 |------|--------------------------------|
 | **왜 필요한가?** | • 카이제곱: "전체적 차이 있음" (어디서?)<br>• 조정 잔차: "이 특정 셀이 기대보다 많음/적음" |
 | **해석 기준** | **±1.96** 초과시 유의한 차이 (p<0.05) |
 | **해석 예시** | 조정잔차 = +2.5 → "이 셀의 관측값이 기대값보다 유의하게 많음"<br>조정잔차 = -2.1 → "이 셀의 관측값이 기대값보다 유의하게 적음" |
 > 💡 **해석 예시**: 남성-A제품 셀의 조정잔차가 +2.5라면 → "남성이 A제품을 선호하는 비율이 예상보다 유의하게 높음"



 ### 5. 효과 크기 vs 잔차 분석

 #### 각각의 역할



 | 구분 | 효과 크기 (Cramér's V) | 조정 잔차 분석 |
 |------|------------------------|---------------|
 | **측정 대상** | **전체 표**의 관계 강도 | **개별 셀**의 차이 |
 | **답하는 질문** | "얼마나 강한 관계인가?" | "구체적으로 어디서 차이나는가?" |
 | **결과 형태** | 0~1 사이 단일 값 | 각 셀별 개별 값들 |

 #### 실제 분석 예시: 성별 × 제품선호도
 **① 카이제곱 검정**: χ² = 25.3, p = 0.001 → "관계 있음"
 **② 효과 크기**: Cramér's V = 0.25 → "중간 정도 관계"
 **③ 잔차 분석**:

 - 남성-A제품: +3.2 (유의하게 많음)

 - 남성-C제품: -2.8 (유의하게 적음)

 - 여성-C제품: +2.8 (유의하게 많음)

 #### 왜 둘 다 필요한가?
 - **효과 크기만**: "관계가 중간 정도다" (구체적 패턴은?)
 - **잔차 분석만**: "이 셀들이 다르다" (전체적 관계 강도는?)
 - **둘 다 사용**: "중간 정도 관계이며, 남성은 A제품을 선호하고 여성은 C제품을 선호하는 패턴"

 💡 **핵심**: 효과 크기는 **숲(전체)**을, 잔차 분석은 **나무(세부)**를 봅니다. 완전한 해석을 위해 둘 다 필요!

 ### 6. 카이제곱검정 수행 체크리스트

 1. **데이터 유형 확인**

    - 범주형 변수인지 확인

    - 연속형이면 구간화 필요


 2. **기대빈도 확인**

    - 모든 셀의 기대빈도 ≥ 5

    - 20% 이상의 셀이 5 미만이면 Fisher's exact test (2x2 표에서만 사용가능)


 3. **검정 수행**

    - 적합도: `chisquare()`

    - 독립성/동질성: `chi2_contingency()`

 4. **결과 해석**

    - p-value < 0.05: 유의한 관계/차이 있음

    - 효과 크기 확인 (Cramér's V) - 자유도별 해석 기준 적용

    - **조정 잔차 분석으로 구체적 차이 파악**

 ### 💡 주의사항

 - **p값 ≠ 효과 크기**: p<0.001이어도 관계는 약할 수 있음

 - **큰 표본 주의**: 표본이 크면 미미한 차이도 유의 → Cramér's V 확인 필수

 - **인과관계 주의**: 카이제곱은 관계성만 확인, 인과관계 증명 X

```python
# 필수 라이브러리 Import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import chi2_contingency, chisquare, fisher_exact
from scipy.stats import chi2
import warnings
import platform

warnings.filterwarnings('ignore')

# 운영체제별 한글 폰트 설정
if platform.system() == 'Windows':
    plt.rcParams['font.family'] = 'Malgun Gothic'
elif platform.system() == 'Darwin':  # macOS
    plt.rcParams['font.family'] = 'AppleGothic'
else:  # Linux
    plt.rcParams['font.family'] = 'NanumGothic'

# 마이너스 기호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False

# 시각화 기본 설정
plt.rcParams['figure.figsize'] = (12, 4)

# 전역 시드 설정
np.random.seed(42)

print("="*50)
print("라이브러리 로드 완료!")
print("한글 폰트 설정 완료!")
print("="*50)
```

  ## 도우미 함수

```python
def cramers_v(chi2_stat, n, r, c):
    """
    Cramér's V 계산
    chi2_stat: 카이제곱 통계량
    n: 전체 표본 수
    r: 행 수
    c: 열 수
    """
    return np.sqrt(chi2_stat / (n * min(r-1, c-1)))

def interpret_cramers_v(v):
    """Cramér's V 해석"""
    if v < 0.1:
        return "매우 약한 관계"
    elif v < 0.3:
        return "약한 관계"
    elif v < 0.5:
        return "중간 관계"
    else:
        return "강한 관계"

def check_expected_frequencies(contingency_table):
    """
    기대빈도 확인
    """
    chi2_stat, p_val, dof, expected = chi2_contingency(contingency_table)
    
    print("\n[기대빈도 확인]")
    print("-"*40)
    
    # 최소 기대빈도
    min_expected = expected.min()
    print(f"최소 기대빈도: {min_expected:.2f}")
    
    # 5 미만인 셀의 비율
    cells_below_5 = (expected < 5).sum()
    total_cells = expected.size
    percent_below_5 = (cells_below_5 / total_cells) * 100
    
    print(f"5 미만 셀: {cells_below_5}/{total_cells} ({percent_below_5:.1f}%)")
    
    if min_expected < 5 or percent_below_5 > 20:
        print("⚠️ 주의: Fisher's exact test 사용 권장")
        return False
    else:
        print("✅ 카이제곱검정 사용 가능")
        return True

def standardized_residuals(observed, expected):
    """표준화 잔차 계산"""
    return (observed - expected) / np.sqrt(expected)
```

  ## 실습 1: 적합도 검정 (Goodness of Fit Test)



  ### 문제: 대학생들의 스터디 카페 이용 시간대 분석

```python
print("\n" + "="*60)
print("실습 1: 스터디 카페 이용 시간대 - 적합도 검정")
print("="*60)

# 스터디 카페 500명 학생의 이용 시간대 데이터
observed_freq = np.array([45, 85, 120, 150, 80, 20])  # 오전, 점심, 오후, 저녁, 밤, 새벽
expected_freq = np.array([83, 83, 83, 83, 84, 84])  # 균등 분포 가정
time_slots = ['오전(6-10)', '점심(10-14)', '오후(14-18)', '저녁(18-22)', '밤(22-2)', '새벽(2-6)']

# DataFrame 생성
study_df = pd.DataFrame({
    '시간대': time_slots,
    '관측빈도': observed_freq,
    '기대빈도': expected_freq,
    '차이': observed_freq - expected_freq
})

print("\n[스터디 카페 이용 시간대 분석]")
display(study_df)
print(f"\n총 이용 학생 수: {observed_freq.sum()}명")
```

```python
# 시각화
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# 막대그래프
x = np.arange(len(time_slots))
width = 0.35

bars1 = axes[0].bar(x - width/2, observed_freq, width, label='관측빈도', alpha=0.8, color='steelblue')
bars2 = axes[0].bar(x + width/2, expected_freq, width, label='기대빈도', alpha=0.8, color='coral')

axes[0].set_xlabel('시간대')
axes[0].set_ylabel('학생 수')
axes[0].set_title('시간대별 스터디 카페 이용 현황')
axes[0].set_xticks(x)
axes[0].set_xticklabels(time_slots, rotation=45, ha='right')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# 원형 그래프
colors = plt.cm.Set3(range(len(time_slots)))
axes[1].pie(observed_freq, labels=time_slots, autopct='%1.1f%%', colors=colors)
axes[1].set_title('시간대별 이용 비율')

# 잔차 플롯
residuals = observed_freq - expected_freq
colors = ['red' if r < 0 else 'green' for r in residuals]
axes[2].bar(time_slots, residuals, color=colors, alpha=0.6)
axes[2].axhline(0, color='black', linestyle='-', linewidth=0.5)
axes[2].set_xlabel('시간대')
axes[2].set_ylabel('잔차 (관측 - 기대)')
axes[2].set_title('잔차 분석')
axes[2].set_xticklabels(time_slots, rotation=45, ha='right')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# 적합도 검정 수행
print("\n[카이제곱 적합도 검정]")
print("-"*40)
print("H₀: 모든 시간대에 균등하게 이용한다")
print("H₁: 특정 시간대에 집중되어 이용한다")
print("유의수준: α = 0.05")

# 카이제곱 검정
chi2_stat, p_value = chisquare(observed_freq, expected_freq)
dof = len(observed_freq) - 1  # 자유도

print(f"\n카이제곱 통계량: {chi2_stat:.4f}")
print(f"자유도: {dof}")
print(f"p-value: {p_value:.4f}")

# 임계값
critical_value = chi2.ppf(0.95, dof)
print(f"임계값 (α=0.05): {critical_value:.4f}")

# 결론
print("\n[결론]")
if p_value < 0.05:
    print(f"✅ p-value({p_value:.4f}) < 0.05 → 귀무가설 기각")
    print("   시간대별로 이용 패턴에 유의한 차이가 있음")
    print("\n[주요 발견]")
    print("   • 오후-저녁 시간대(14-22시) 집중 이용")
    print("   • 새벽 시간대(2-6시) 이용률 매우 낮음")
else:
    print(f"❌ p-value({p_value:.4f}) ≥ 0.05 → 귀무가설 채택")
    print("   시간대별 이용이 균등하다고 볼 수 있음")
```

  ## 실습 2: 독립성 검정 (Test of Independence)



  ### 문제: 학년별 프로그래밍 언어 선호도 분석

```python
print("\n" + "="*60)
print("실습 2: 학년별 프로그래밍 언어 선호도 - 독립성 검정")
print("="*60)

# 데이터 생성: 학년 × 선호 언어
data = {
    'Python': [55, 45, 25],  # [2학년, 3학년, 4학년]
    'Java': [20, 35, 40],
    'JavaScript': [25, 20, 35]
}

contingency_table = pd.DataFrame(data, index=['2학년', '3학년', '4학년'])
print("\n[분할표 (Contingency Table)]")
display(contingency_table)
```

```python
# 시각화
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# 누적 막대그래프
contingency_table.T.plot(kind='bar', stacked=True, ax=axes[0], color=['#3498db', '#e74c3c', '#2ecc71'])
axes[0].set_title('언어별 학년 분포')
axes[0].set_xlabel('프로그래밍 언어')
axes[0].set_ylabel('학생 수')
axes[0].legend(title='학년')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=0)

# 히트맵
sns.heatmap(contingency_table, annot=True, fmt='d', cmap='YlOrRd', ax=axes[1], cbar_kws={'label': '학생 수'})
axes[1].set_title('학년 × 언어 선호도 히트맵')

# 비율 막대그래프
prop_table = contingency_table.div(contingency_table.sum(axis=1), axis=0) * 100
prop_table.plot(kind='bar', ax=axes[2], color=['#9b59b6', '#f39c12', '#1abc9c'])
axes[2].set_title('학년별 언어 선호 비율(%)')
axes[2].set_ylabel('비율(%)')
axes[2].set_xticklabels(axes[2].get_xticklabels(), rotation=0)
axes[2].legend(title='언어', bbox_to_anchor=(1.05, 1))

plt.tight_layout()
plt.show()
```

```python
# 기대빈도 확인
is_valid = check_expected_frequencies(contingency_table)
```

```python
# 독립성 검정 수행
print("\n[카이제곱 독립성 검정]")
print("-"*40)
print("H₀: 학년과 프로그래밍 언어 선호도는 독립적이다")
print("H₁: 학년과 프로그래밍 언어 선호도는 관련이 있다")
print("유의수준: α = 0.05")

# 카이제곱 검정
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"\n카이제곱 통계량: {chi2_stat:.4f}")
print(f"자유도: {dof}")
print(f"p-value: {p_value:.4f}")

# Cramér's V
n = contingency_table.values.sum()
r, c = contingency_table.shape
v = cramers_v(chi2_stat, n, r, c)
print(f"Cramér's V: {v:.3f} ({interpret_cramers_v(v)})")

# 기대빈도 테이블
expected_df = pd.DataFrame(expected, 
                           index=contingency_table.index,
                           columns=contingency_table.columns)
print("\n[기대빈도]")
display(expected_df.round(2))

# 표준화 잔차
std_residuals = standardized_residuals(contingency_table.values, expected)
residuals_df = pd.DataFrame(std_residuals,
                            index=contingency_table.index,
                            columns=contingency_table.columns)
print("\n[표준화 잔차]")
display(residuals_df.round(2))
print("(|잔차| > 2: 유의한 차이)")

# 결론
print("\n[결론]")
if p_value < 0.05:
    print(f"✅ p-value({p_value:.4f}) < 0.05 → 귀무가설 기각")
    print("   학년과 프로그래밍 언어 선호도는 관련이 있음")
    print(f"   효과 크기: {interpret_cramers_v(v)}")
    
    print("\n[구체적 패턴 분석]")
    for i, grade in enumerate(contingency_table.index):
        for j, lang in enumerate(contingency_table.columns):
            if abs(std_residuals[i, j]) > 2:
                if std_residuals[i, j] > 0:
                    print(f"   • {grade} - {lang}: 예상보다 선호도 높음")
                else:
                    print(f"   • {grade} - {lang}: 예상보다 선호도 낮음")
else:
    print(f"❌ p-value({p_value:.4f}) ≥ 0.05 → 귀무가설 채택")
    print("   학년과 프로그래밍 언어 선호도는 독립적임")
```

  ## 실습 3: 동질성 검정 (Test of Homogeneity)



  ### 문제: 세 지역의 정치 성향 비교

```python
print("\n" + "="*60)
print("실습 3: 지역별 정치 성향 - 동질성 검정")
print("="*60)

# 데이터: 지역별 정당 지지 (차이가 거의 없도록 설정)
data = {
    '진보': [102, 98, 100],
    '중도': [97, 103, 100],
    '보수': [101, 99, 100]
}

regions = ['서울', '부산', '대구']
region_df = pd.DataFrame(data, index=regions)

print("\n[지역별 정당 지지 현황]")
print(region_df)

# 비율 계산
prop_df = region_df.div(region_df.sum(axis=1), axis=0) * 100
print("\n[지역별 지지 비율(%)]")
display(prop_df.round(1))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# 지역별 분포
region_df.plot(kind='bar', ax=axes[0], color=['#3498db', '#e67e22', '#27ae60'])
axes[0].set_title('지역별 정당 지지 분포')
axes[0].set_xlabel('지역')
axes[0].set_ylabel('지지자 수')
axes[0].legend(title='정당')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=0)

# 100% 누적 막대그래프
prop_df.plot(kind='bar', stacked=True, ax=axes[1], color=['#9b59b6', '#f39c12', '#1abc9c'])
axes[1].set_title('지역별 정당 지지 비율')
axes[1].set_xlabel('지역')
axes[1].set_ylabel('비율(%)')
axes[1].set_ylim(0, 100)
axes[1].legend(title='정당', bbox_to_anchor=(1.05, 1))
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=0)

# 히트맵
sns.heatmap(region_df, annot=True, fmt='d', cmap='RdBu_r', ax=axes[2], cbar_kws={'label': '지지자 수'})
axes[2].set_title('지역 × 정당 히트맵')

plt.tight_layout()
plt.show()
```

```python
print("\n[카이제곱 동질성 검정]")
print("-"*40)
print("H₀: 세 지역의 정치 성향 분포는 동일하다")
print("H₁: 적어도 한 지역의 정치 성향 분포가 다르다")
print("유의수준: α = 0.05")

# 기대빈도 확인
is_valid = check_expected_frequencies(region_df)

# 카이제곱 검정
chi2_stat, p_value, dof, expected = chi2_contingency(region_df)

print(f"\n카이제곱 통계량: {chi2_stat:.4f}")
print(f"자유도: {dof}")
print(f"p-value: {p_value:.4f}")

# Cramér's V
n = region_df.values.sum()
r, c = region_df.shape
v = cramers_v(chi2_stat, n, r, c)
print(f"Cramér's V: {v:.3f} ({interpret_cramers_v(v)})")

# 표준화 잔차 분석
std_residuals = standardized_residuals(region_df.values, expected)
residuals_df = pd.DataFrame(std_residuals,
                            index=region_df.index,
                            columns=region_df.columns)

print("\n[표준화 잔차]")
print(residuals_df.round(2))

# 유의한 차이 찾기
print("\n[유의한 차이 (|잔차| > 2)]")
significant = np.abs(std_residuals) > 2
has_significant = False
for i, region in enumerate(regions):
    for j, party in enumerate(region_df.columns):
        if significant[i, j]:
            has_significant = True
            residual = std_residuals[i, j]
            if residual > 0:
                print(f"• {region} - {party}: 기대보다 많음 (잔차={residual:.2f})")
            else:
                print(f"• {region} - {party}: 기대보다 적음 (잔차={residual:.2f})")

if not has_significant:
    print("• 유의한 차이가 있는 셀이 없음")

# 결론
print("\n[결론]")
if p_value < 0.05:
    print(f"✅ p-value({p_value:.4f}) < 0.05 → 귀무가설 기각")
    print("   지역별 정치 성향에 유의한 차이가 있음")
    print(f"   효과 크기: {interpret_cramers_v(v)}")
else:
    print(f"❌ p-value({p_value:.4f}) ≥ 0.05 → 귀무가설 채택")
    print("   지역별 정치 성향이 동일하다고 볼 수 있음")
    print("\n[해석]")
    print("   • 세 지역 모두 정당 지지 비율이 거의 균등함 (각 약 33%)")
    print("   • 지역 간 정치 성향의 차이가 통계적으로 유의하지 않음")
    print("   • 표본에서 지역별 특성이 나타나지 않음")
```

  ## 실습 4: Fisher's Exact Test (작은 표본)



  ### 문제: 새로운 교육 방법의 효과 검정 (2×2 표)

```python
print("\n" + "="*60)
print("실습 4: 교육 방법 효과 - Fisher's Exact Test")
print("="*60)

# 2x2 분할표: 시험 통과 여부
education_table = pd.DataFrame({
    '통과': [9, 3],
    '미통과': [2, 6]
}, index=['신규교육', '기존교육'])

print("\n[교육 효과 분할표]")
display(education_table)
```

```python
# 시각화
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# 비율 막대그래프
prop_table = education_table.div(education_table.sum(axis=1), axis=0) * 100
prop_table.plot(kind='bar', ax=axes[0], color=['#2ecc71', '#e74c3c'])
axes[0].set_title('교육 방법별 통과율(%)')
axes[0].set_ylabel('비율(%)')
axes[0].set_xlabel('교육 방법')
axes[0].legend(title='결과')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=0)

# 모자이크 플롯 (간단한 버전)
from matplotlib.patches import Rectangle

ax = axes[1]
total = education_table.values.sum()
x_pos = 0

for i, method in enumerate(education_table.index):
    y_pos = 0
    row_sum = education_table.iloc[i].sum()
    width = row_sum / total
    
    for j, outcome in enumerate(education_table.columns):
        value = education_table.iloc[i, j]
        height = value / row_sum
        
        color = '#2ecc71' if j == 0 else '#e74c3c'
        rect = Rectangle((x_pos, y_pos), width, height, 
                        facecolor=color, edgecolor='black', linewidth=1, alpha=0.7)
        ax.add_patch(rect)
        
        # 레이블 추가
        ax.text(x_pos + width/2, y_pos + height/2, 
               f'{value}', ha='center', va='center', fontsize=12, fontweight='bold')
        
        y_pos += height
    
    ax.text(x_pos + width/2, -0.05, method, ha='center', fontsize=11)
    x_pos += width

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_title('모자이크 플롯')
ax.set_ylabel('결과 비율')
ax.axis('off')

plt.tight_layout()
plt.show()

```

```python
# Fisher's Exact Test
print("\n[Fisher's Exact Test]")
print("-"*40)
print("H₀: 교육 방법과 통과 여부는 독립적이다")
print("H₁: 교육 방법과 통과 여부는 관련이 있다")
print("유의수준: α = 0.05")

# 기대빈도 확인
print("\n[기대빈도 확인]")
chi2_stat_check, _, _, expected = chi2_contingency(education_table)
print("기대빈도:")
print(pd.DataFrame(expected, 
                   index=education_table.index,
                   columns=education_table.columns).round(2))

min_expected = expected.min()
if min_expected < 5:
    print(f"⚠️ 최소 기대빈도 = {min_expected:.2f} < 5")
    print("   → Fisher's Exact Test 사용")

# Fisher's exact test
oddsratio, p_value = fisher_exact(education_table)

print(f"\n[검정 결과]")
print(f"Odds Ratio: {oddsratio:.3f}")
print(f"p-value: {p_value:.4f}")

# Odds Ratio 해석
# 오즈 = 통과자수/미통과자수 (분할표의 실제 빈도 사용)

# 2×2 분할표에서 오즈비 계산:
#           통과    미통과
# 신규교육    a        b     → 신규 오즈 = a/b
# 기존교육    c        d     → 기존 오즈 = c/d
# 오즈비(OR) = (a/b)/(c/d) = (a×d)/(b×c)

# 💡 해석 예시:
# OR = 2.5 → "신규교육 받은 사람이 통과할 가능성이 기존교육보다 2.5배 높음"
# OR = 0.4 → "신규교육 받은 사람이 통과할 가능성이 기존교육의 40%에 불과"

print(f"\n[Odds Ratio 해석]")
if oddsratio > 1:
    print(f"신규교육의 통과 오즈가 기존교육보다 {oddsratio:.1f}배 높음")
elif oddsratio < 1:
    print(f"신규교육의 통과 오즈가 기존교육보다 {1/oddsratio:.1f}배 낮음")
else:
    print("신규교육과 기존교육의 통과 오즈가 동일")

# 결론
print("\n[결론]")
if p_value < 0.05:
    print(f"✅ p-value({p_value:.4f}) < 0.05 → 귀무가설 기각")
    print("   교육 방법과 통과 여부는 관련이 있음")
    print("   신규교육 방법이 더 효과적임")
else:
    print(f"❌ p-value({p_value:.4f}) ≥ 0.05 → 귀무가설 채택")
    print("   교육 방법과 통과 여부는 독립적임")
```
