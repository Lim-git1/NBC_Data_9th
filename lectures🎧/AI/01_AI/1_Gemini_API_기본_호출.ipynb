### 1. 기본 텍스트 생성 (Single-turn Conversation)

[Gemini API 문서 링크](https://ai.google.dev/docs)

(1) Single-turn이란?
- **Single-turn**: 한 번의 질문과 한 번의 응답으로 완결되는 대화 방식
- 이전 대화 맥락(context)을 기억하지 않음
- 독립적인 질의응답에 적합 (예: 번역, 요약, 간단한 질문)
- 반대 개념: **Multi-turn** (여러 차례 주고받는 대화, 문맥 유지)

(2) Google GenAI SDK를 통한 Single-turn 호출
- **SDK(Software Development Kit)** 는 특정 플랫폼이나 서비스를 쉽게 사용할 수 있도록 만들어진 개발 도구 모음입니다.

````python
from google import genai

# 클라이언트 초기화
client = genai.Client(api_key=api_key)

# 기본 텍스트 생성 요청
search_query = "google gemini LLM의 특징은 무엇인가요?"

response = client.models.generate_content(
    model=gemini_model,      # 사용할 AI 모델 (예: gemini-2.0-flash)
    contents=search_query    # AI에게 보낼 질문/명령 (프롬프트)
)

# 응답 출력
if response:
    print(f"질문: {search_query}")
    print(f"답변: {response.text}")
    print("-" * 50)
else:
    print("응답을 받지 못했습니다.")
````

(2) 매개변수를 활용한 상세 설정
- [API 파라미터 문서 링크](https://ai.google.dev/api/generate-content?hl=ko#v1beta.GenerationConfig)


````python
from google.genai import types
from pprint import pprint 

# GenerateContentConfig: 생성 모델의 동작 방식을 세밀하게 제어하는 설정
    
    #temperature
    # -> 값이 커지면 단어별 확률이 비슷해짐 -> 출력 단어 후보군이 생김 -> 정확도가 좀 떨어짐 but 창의적
    # top_p
    # -> 후보군의 개수를 정하는 파라미터
    # top_k
    # max_output_tokens=
    # candidate_count
    # system_instruction
    
generation_config = types.GenerateContentConfig(
    temperature=0.7,       # 기본값: 1.0 (창의성 조절, 0.0~2.0)
    top_p=0.95,            # 기본값: 0.95 (다양성 조절, 0.0~1.0)
    top_k=64,              # 토큰 선택 범위
    max_output_tokens=200, # 최대 출력 토큰 수: 답변 길이 조절
    candidate_count=2,     # 기본값: 1 (응답 생성 수) 동일한 질문에 답변을 많이 받고싶으면 숫자 늘려 but 토큰이 쓰이죠
    system_instruction="너는 AI 기술을 쉽게 설명하는 전문가야. 초보자도 이해할 수 있게 설명해줘."
)

# 질문 설정 - 명확하고 구체적인 프롬프트 예시
question = """
LLM(Large Language Model)과 VLM(Vision Language Model)의 차이점을 설명해주세요.

조건:
- 각 모델이 무엇인지 한 문장으로 정의
- 주요 차이점 3가지를 명확히 구분
- 각각의 실제 활용 예시 1개씩 포함
- 전체 5문장 이내로 작성
"""

try:
    response = client.models.generate_content(
        model=gemini_model,
        contents=question,
        config=generation_config
    )
    
    print(f"질문: {question}")
    print(f"모델: gemini-2.0-flash")
    print(f"설정: 창의성={generation_config.temperature}, 다양성={generation_config.top_p}")
    print(f"생성된 응답 수: {len(response.candidates)}")
    print(f"생성된 응답(원본): ")
    pprint(response)
    print("=" * 80)
    
    # 모든 응답 출력
    for i, candidate in enumerate(response.candidates, 1):
        print(f"\n📝 응답 {i}:")
        print("-" * 50)
        pprint(candidate.content.parts[0].text)
        print(f"🏁 종료 이유: {candidate.finish_reason}")
        if i < len(response.candidates):
            print("=" * 80)
    
except Exception as e:
    print(f"API 호출 오류: {e}")
````

(3) 응답 토큰 수 확인 및 비용 산출
- [Gemini 모델 비용 문서 링크](https://ai.google.dev/gemini-api/docs/pricing?hl=ko)

실무(최종프로젝트)시 비용 계산 필요성


ex. 
- 뉴스기사 1000건을 분석히는데 비용이 얼마나 들지 추정

- 1개의 기사당 평균적으로 단어1000 -> 1000개의 토큰 * 비용계산 * 기사 수

비용계산 지피티한테 시키지 말 것.(지피티 숫자계산 못함)

**공식문서들어기서 확인하고 직접 계산하는 것이 best**

````python
# 토큰 사용량 정보
if response.usage_metadata:
    
    prompt_tokens = response.usage_metadata.prompt_token_count
    output_tokens = response.usage_metadata.candidates_token_count
    total_tokens = response.usage_metadata.total_token_count
        
    # gemini-2.5-flash-lite 무료 등급 가격 (백만 토큰당)
    input_price = 0.10
    output_price = 0.40
    exchange_rate = 1500  # 환율
    
    # 비용 계산 (USD)
    input_cost_usd = (prompt_tokens / 1_000_000) * input_price
    output_cost_usd = (output_tokens / 1_000_000) * output_price
    total_cost_usd = input_cost_usd + output_cost_usd
    
    # 비용 계산 (KRW)
    input_cost_krw = input_cost_usd * exchange_rate
    output_cost_krw = output_cost_usd * exchange_rate
    total_cost_krw = total_cost_usd * exchange_rate
    
    print(f"\n" + "=" * 60)
    print(f"{'토큰 사용량 및 비용':^60}")
    print("=" * 60)
    print(f"프롬프트 토큰: {prompt_tokens:>6,} tokens")
    print(f"              ${input_cost_usd:>9.6f}  (₩{input_cost_krw:>7.4f})")
    print(f"-" * 60)
    print(f"응답 토큰:     {output_tokens:>6,} tokens")
    print(f"              ${output_cost_usd:>9.6f}  (₩{output_cost_krw:>7.4f})")
    print(f"-" * 60)
    print(f"총 토큰:       {total_tokens:>6,} tokens")
    print(f"총 비용:      ${total_cost_usd:>9.6f}  (₩{total_cost_krw:>7.4f})")
    print("=" * 60)
````

### 2. 스트리밍 응답

**일반 응답 방식 (기본)**
- AI가 답변을 **완전히 다 만든 후**에 한 번에 보여줍니다
- 긴 답변일수록 기다리는 시간이 길어집니다

**스트리밍 응답 방식 (개선)**
- AI가 답변을 **만드는 동시에 조금씩** 보여줍니다
- 사용자가 기다림 없이 답변을 바로 읽을 수 있어 더 자연스럽습니다



> 스트림릿에서 ai 연동해서 서비스 가능. 위에 방식도 가능하고 이 방식도 가능한데,
> 
> 얘가 좀 더 까다롭긴하지ㅏㄴ 얘가 실시간이라 더 적합함

````python
question = "AI가 어떻게 작동하는지 간단히 설명해주세요"
print(f"질문: {question}")
print("\n실시간 응답:")

response = client.models.generate_content_stream(
    model=gemini_model,
    contents=[question]
)

full_response = ""
for chunk in response:
    if chunk.text:
        print(chunk.text, end="", flush=True)  # 실시간 출력
        full_response += chunk.text
        
print(f"\n\n📝 전체 응답 길이: {len(full_response)} 문자")
````

### 3. 대화형 채팅 구현
(1) 멀티 턴(Multi-turn) 기본 대화 시스템


**Gemini 기본 Role 종류**
Gemini API에서 사용하는 주요 role들:
```python
# 1. user - 사용자의 입력
{"role": "user", "content": "안녕하세요!"}

# 2. model - Gemini의 응답 (openai의 assistant 대신 사용)
{"role": "model", "content": "안녕하세요! 어떻게 도와드릴까요?"}

```

````python
# 하이퍼파라미터 설정
generation_config = types.GenerateContentConfig(
   temperature=0.8,        # 창의성 조절 (0.0~2.0)
   top_p=0.9,             # 다양성 조절 (0.0~1.0)
   top_k=40,              # 토큰 선택 범위
   max_output_tokens=500, # 최대 출력 토큰 수
   candidate_count=1,      # 응답 생성 수
   system_instruction="너는 데이터 분석 분야의 시니어 전문가야. 실무 경험을 바탕으로 구체적이고 실용적인 조언을 제공해."
)

# 새 채팅 세션 생성
chat = client.chats.create(model=gemini_model, config=generation_config)

# 첫 번째 메시지
print("\n💬 첫 번째 대화:")
user_msg1 = "저는 지금 데이터 분석 직무에 취업을 하려해요."
print(f"👤 사용자: {user_msg1}")

response1 = chat.send_message(user_msg1)
display(f"🤖 AI 전문가: {response1.text}")

# 두 번째 메시지 (컨텍스트를 기억함)
print(f"\n💬 두 번째 대화:")
user_msg2 = "제가 가장 먼저 무엇을 준비하면 좋을까요?"
print(f"👤 사용자: {user_msg2}")

response2 = chat.send_message(user_msg2)
display(f"🤖 AI 전문가: {response2.text}")

# 세 번째 메시지 (더 구체적인 질문)
print(f"\n💬 세 번째 대화:")
user_msg3 = "Python은 어느 정도 수준까지 알아야 하나요?"
print(f"👤 사용자: {user_msg3}")

response3 = chat.send_message(user_msg3)
display(f"🤖 AI 전문가: {response3.text}")
````

````python
# 모든 대화 기록 얻기
chat.get_history()
````

(2) 멀티턴 대화 + 스트리밍 구현

````python
# 하이퍼파라미터 설정
generation_config = types.GenerateContentConfig(
   temperature=0.8,
   top_p=0.9,
   top_k=40,
   max_output_tokens=300,
   candidate_count=1,
   system_instruction="너는 친근하고 도움이 되는 AI 어시스턴트야. 사용자와 자연스럽게 대화해."
)


print("🤖 실시간 Gemini 채팅 (스트리밍)")
print("=" * 60)
print("💡 명령어:")
print("  • '종료' 또는 'quit' : 채팅 종료")
print("=" * 60)

# 새 채팅 세션 생성
chat_session = client.chats.create(
    model=gemini_model,
    config=generation_config
)

turn_count = 0  # 대화 턴 수

# 실시간 대화 루프
while True:
    # 사용자 입력 받기
    print(f"\n💭 대화 {turn_count + 1}")
    user_message = input("👤 사용자: ").strip()
    
    # 종료 명령 확인
    if user_message.lower() in ['quit', 'exit', '종료', 'q']:
        print("\n👋 채팅을 종료합니다!")
        break
    
    # 빈 입력 처리
    if not user_message:
        print("❓ 메시지를 입력해주세요.")
        continue
    
    # 사용자 메시지 출력
    print(f"👤 사용자: {user_message}")
    
    # AI 응답 시작
    print("🤖 AI: ", end="", flush=True)
    
    # 스트리밍으로 응답 받기
    streaming_response = chat_session.send_message_stream(user_message)

    # 실시간 스트리밍 출력
    for chunk in streaming_response:
        if chunk.text:
            print(chunk.text, end="", flush=True)

    print()  # 줄바꿈
    turn_count += 1
````
